Train the yolo model

```python
start = Path.cwd().resolve()
BASE_DIR = next(
    (directory for directory in (start, *start.parents)
     if (directory / "src" / "cnn_lstm_pytorch.py").is_file()),
    None,
)
if BASE_DIR is None:
    raise FileNotFoundError("Start the notebook kernel inside the violence-movies project.")
print(f"Base directory: {BASE_DIR}")
```

```python
start = Path.cwd().resolve()
BASE_DIR = next(
    (directory for directory in (start, *start.parents)
     if (directory / "src" / "cnn_lstm_pytorch.py").is_file()),
    None,
)
if BASE_DIR is None:
    raise FileNotFoundError("Start the notebook kernel inside the violence-movies project.")

data_path = BASE_DIR / "data/roboflow/Violence-Detection-3/data.yaml"
yolo_weight_path = BASE_DIR / "yolo_weight"

model = YOLO(yolo_weight_path / "yolov8s.pt")
results = model.train(
    data=data_path,
    epochs=120,
    imgsz=640,
    batch=24,
    device=0,
    workers=8,
    project=str(BASE_DIR / "src" / "runs" / "detect"),
    name='train_yolov8s_' + "violence" + "_v3",
    verbose=True
)
```
